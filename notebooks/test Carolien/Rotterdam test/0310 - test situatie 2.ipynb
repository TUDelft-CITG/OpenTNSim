```python
required_depth = vessel_draught * (1 + ukc_p[0])

dt = (HvH_data.index[1] - HvH_data.index[0]).total_seconds() / 3600

joint_access_hours = []


botlek_data['Water level'] = botlek_data['Water level'] / 100

for i in range(len(low_tides)-1):

    start = low_tides.index[i]
    end = low_tides.index[i+1]

    duration = (end - start).total_seconds()/3600

    cycle_HvH = HvH_data.loc[start:end].copy()
    #cycle_Rot = Rotterdam_data.loc[start:end].copy()
    cycle_bot = botlek_data.loc[start:end].copy()

    cycle_HvH["hours_since_low"] = (cycle_HvH.index - start).total_seconds()/3600
    cycle_bot["hours_since_low"] = (cycle_bot.index - start).total_seconds()/3600

    hours = cycle_HvH["hours_since_low"]

    depth_HvH = cycle_HvH['Water level'] + waterway_depth
    depth_bot = cycle_bot['Water level'] + waterway_depth

    joint_access = (depth_HvH >= required_depth) & (depth_bot >= required_depth)
    joint_hours = joint_access.sum() * dt
    joint_access_hours.append(joint_hours)
    
    plt.figure(figsize=(6,3))

    plt.plot(cycle_HvH["hours_since_low"], depth_HvH, label="Hoek van Holland")
    plt.plot(cycle_bot["hours_since_low"], depth_bot, label="Botlek")

    plt.axhline(required_depth, color='red', linestyle='--', label="Required depth")

    plt.fill_between(hours,
                     required_depth,
                     depth_HvH,
                     where=joint_access,
                     color='green',
                     alpha=0.3,
                     label="Joint accessibility")

    plt.xlabel("Hours since low tide (HvH)")
    plt.ylabel("Available depth [m]")
    plt.title(f"Tidal cycle {i+1}\n{start} → {end:} | Duration: {duration:.2f} h")

    plt.figtext(0.5, -0.10,
            f"Tidal accessibility over waterway length: {joint_hours:.2f} hours",
            ha="center")

    
    plt.legend()
    plt.show()
```

```python
# make your preferred Vessel class out of available mix-ins.
Vessel = type(
    "Vessel", 
    (HasPortAccess,
     HasTerminal,
     PassesAnchorage,
     HasDraughtRestrictions,
     HasMultiDiGraph,
     Identifiable, # allows to give the object a name and a random ID,
     Movable,      # allows the object to move, with a fixed speed, while logging this activity
     VesselProperties,
    ), 
    {}
)
```

```python
def mission(env, vessel):
    """
    Method that defines the mission of the vessel. 
    
    In this case: 
        keep moving along the path until its end point is reached
    """
    yield from vessel.move()
```

```python
# create vessel from a dict 
path = nx.dijkstra_path(env.graph, 'Hoek van Holland', 'Rotterdam' )
data_vessel = {
    "env": env,                                       # needed for simpy simulation
    "name": "Vessel 1",                               # required by Identifiable
    "geometry": env.graph.nodes[path[0]]['geometry'], # required by Locatable
    "route": path,                                    # required by Routeable
    "v": vessel_speed,                                # required by Movable, 1 m/s to check if the distance is covered in the expected time
    "bound": 'inbound',
    "terminal": terminal,
    "berthing_time":15, #minutes
    "loading_time": 12, #hours
    "deberthing_time":5, #minutes
    "next_destination": 'Hoek van Holland',
    "type":"tanker",
    "L":vessel_length,
    "B":vessel_beam,
    "T":vessel_draught,
}  # 

# create an instance of the Vessel class using the input dict data_vessel
vessel = Vessel(**data_vessel)

# start the simulation
vessel.mission = env.process(mission(env, vessel))
```
