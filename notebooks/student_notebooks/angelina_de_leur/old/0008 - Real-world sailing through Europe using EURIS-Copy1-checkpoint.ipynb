# Generate a vessel
Create a vessel class using the movable and identifiable mixins. 

```python
# make your preferred Vessel class out of available mix-ins.
Vessel = type(
    "Vessel", 
    (
        opentnsim.core.Identifiable, # allows to give the object a name and a random ID,
        opentnsim.core.Movable,      # allows the object to move, with a fixed speed, while logging this activity
    ), 
    {}
)
```

```python
maasvlakte = shapely.Point(4.0566946, 51.9471624)
#duisburg = shapely.Point(6.722793, 51.451388)
kreekrak = shapely.Point(4.228202, 51.461103)
```

```python
nodes_gdf = gpd.GeoDataFrame(graph.nodes.values(), index=graph.nodes.keys())
```

```python
distances, idx = nodes_gdf.sindex.nearest(maasvlakte)
maasvlakte_node = nodes_gdf.iloc[idx[0]]

distances, idx = nodes_gdf.sindex.nearest(kreekrak)
kreekrak_node = nodes_gdf.iloc[idx[0]]
```

```python
route = nx.shortest_path(graph, maasvlakte_node.name, kreekrak_node.name, weight='length_m')
```

# Show the graph
Show the graph and selected route on a map. 

```python
# Create a map centered between the two points
m = folium.Map(location=[51.83, 4.33], zoom_start = 10, tiles="cartodb positron")

for edge in graph.edges(data = True):
    points_x = list(edge[2]["geometry"].coords.xy[0])
    points_y = list(edge[2]["geometry"].coords.xy[1])
    
    line = []
    for i, _ in enumerate(points_x):
        line.append((points_y[i], points_x[i]))

    if edge[0] in route and edge[1] in route:
        folium.PolyLine(line, color = "red", weight = 3, popup = edge[2].get("name", "")).add_to(m)
    else:
        folium.PolyLine(
            line, 
            color="black", 
            weight=3, 
            opacity=0.1,
            popup=edge[2].get("name", "")
        ).add_to(m)

for node in graph.nodes(data = True):
    point = list(node[1]["geometry"].coords.xy)
    folium.CircleMarker(
        location=[point[1][0],point[0][0]], 
        color='black',
        fill_color = "black", 
        fill=True, 
        radus = 1, 
        opacity = 0.1,
        popup = node[0]
    ).add_to(m)
        
# Add round marker for Maasvlakte with popup
folium.CircleMarker(
    location=[maasvlakte_node.geometry.y, maasvlakte_node.geometry.x],
    radius=8,
    color='green',
    fill=True,
    fill_color='green',
    fill_opacity=0.7,
    popup='Maasvlakte'
).add_to(m)

# Add round marker for Kreekrak with popup
folium.CircleMarker(
    location=[kreekrak_node.geometry.y, kreekrak_node.geometry.x],
    radius=8,
    color='blue',
    fill=True,
    fill_color='blue',
    fill_opacity=0.7,
    popup='Moerdijk'
).add_to(m)

# Display the map
m
```

# Define a move process
Keep moving until the vessel has reached the destination. 

```python
def mission(env, vessel):
    """
    Method that defines the mission of the vessel. 
    
    In this case: 
        keep moving along the path until its end point is reached
    """
    while True:
        yield from vessel.move()
        
        if vessel.geometry == nx.get_node_attributes(env.graph, "geometry")[vessel.route[-1]]:
            break
```

# Run a simulation
Run a simple simulation using one vessel. 

```python
# start simpy environment
simulation_start = datetime.datetime(2024, 1, 1, 0, 0, 0)
env = simpy.Environment(initial_time=simulation_start.timestamp())
env.epoch = simulation_start

# add graph to environment
env.graph = graph

# create vessel from a dict 
path = route
data_vessel = {
    "env": env,                                       # needed for simpy simulation
    "name": "Vessel",                                 # required by Identifiable
    "geometry": env.graph.nodes[path[0]]['geometry'], # required by Locatable
    "route": path,                                    # required by Routeable
    "v": 1,                                           # required by Movable, 1 m/s to check if the distance is covered in the expected time
}  # 

# create an instance of the Vessel class using the input dict data_vessel
vessel = Vessel(**data_vessel)

# start the simulation
env.process(mission(env, vessel))
env.run()
```
