```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

fig, ax_v = plt.subplots(1, 1, figsize=(10, 3), sharex=True)

# Empty plot limits
ax_v.set_xlim(-100, 2600)
ax_v.set_ylim(-0.2, 6.2)

# Labels and title
ax_v.set_ylabel("Velocity (m/s)")
ax_v.set_xlabel("Time after start lock passage (s)")
ax_v.set_title("Lock passage")
ax_v.grid(True)

# Fake legend
legend_lines = [
    Line2D([0], [0], color="tab:green", linewidth=2, label="van der Werff (2025)"),
    Line2D([0], [0], color="tab:orange", linewidth=2, label="Segers (2020)")
]

ax_v.legend(handles=legend_lines, loc="upper right")

plt.tight_layout()
plt.show()
```
