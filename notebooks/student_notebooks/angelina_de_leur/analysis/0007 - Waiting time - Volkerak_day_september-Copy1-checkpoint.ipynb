#### 2. Load, read and filter outliers of speed data 

For this example we'll use a representative AIS data file called "ais.csv". Make sure to have this file in a known (and convenient) directory. We'll use the combination of pandas, geopandas and movingpandas to read the data.
Pandas reads the data as a table. Geopandas adds the location and movingpandas adds the time information. 
The result is that we can use data as table, feature (geospatial) and trajectory (moving objects). 

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

```python
# Determine where your data file is and create the directory
data_dir = pathlib.Path('./data') 
```

```python
ais_df = pd.read_csv(data_dir / 'Volkerak_day_sept.csv')
# ais_df = pd.read_csv(data_dir / 'Volkerak_week_sept.csv')
ais_df['geometry'] = ais_df.apply(lambda x: shapely.geometry.Point(x['longitude_deg'], x['latitude_deg']), axis=1)
```

```python
# Check outliers 
speed = ais_df['speedOverGround_kts']
speed_max = ais_df['speedOverGround_kts'].max()

top10 = ais_df.nlargest(10, 'speedOverGround_kts')
mean_speed = ais_df['speedOverGround_kts'].mean()


# print(top10[['mmsi', 'speedOverGround_kts']])
print(top10[['mmsi', 'speedOverGround_kts', 'calcSpeedOverGround_m_s']])
print(f"Max speed in selected data: {speed_max}")
print(f"Mean speed selected data: {mean_speed:.2f} kts")
```

```python
# Remove outliers using the 99th percentile
speed_threshold = ais_df['speedOverGround_kts'].quantile(0.99)
ais_clean = ais_df[ais_df['speedOverGround_kts'] <= speed_threshold]

print("Messages excluded:", len(ais_df) - len(ais_clean))

max_speed = ais_clean['speedOverGround_kts'].max()
mean_speed = ais_clean['speedOverGround_kts'].mean()

print(f"Max speed (cleaned): {max_speed:.2f} kts")
print(f"Mean speed (cleaned): {mean_speed:.2f} kts")
```

```python
# Create figure with 1 row and 2 columns
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Plot 1: speedOverGround_kts ---
axes[0].hist(ais_df['speedOverGround_kts'], bins=60,
             color='steelblue', edgecolor='black')
axes[0].set_title("Speed Over Ground (kts)")
axes[0].set_xlabel("Speed (kts)")
axes[0].set_ylabel("Count")
axes[0].set_ylim(0,500) #ylim
axes[0].grid(alpha=0.3)

# --- Plot 2: calcSpeedOverGround_m_s ---
axes[1].hist(ais_df['calcSpeedOverGround_m_s'], bins=60,
             color='darkorange', edgecolor='black')
axes[1].set_title("Calculated Speed Over Ground (m/s)")
axes[1].set_xlabel("Speed (m/s)")
axes[1].set_ylabel("Count")
axes[1].set_ylim(0,500) # ylim
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Ensure geometry is parsed correctly
if isinstance(ais_df['geometry'].iloc[0], str):
    ais_df['geometry'] = ais_df['geometry'].apply(wkt.loads)

# Create folium map centered on average coordinates
map_center = [ais_df['latitude_deg'].mean() + 0.1, ais_df['longitude_deg'].mean() - 0.35]
m = folium.Map(location=map_center, zoom_start=10)

# Normalize speed for color mapping 
# norm = colors.Normalize(vmin=ais_df['speedOverGround_kts'].min(), vmax=ais_df['speedOverGround_kts'].max())
norm = colors.Normalize(vmin=ais_df['speedOverGround_kts'].min(), vmax=max_speed)
colormap = plt.get_cmap('viridis')

# Add points to the map with name and speed in popup
for _, row in ais_df.iterrows():
    if pd.notnull(row['geometry']):# and pd.notnull(row['SPEED']):

        lat, lon = row['geometry'].y, row['geometry'].x
        speed = row['speedOverGround_kts']
        # name = row['mmsi'] if pd.notnull(row['name']) else 'Unknown'
        name = row['mmsi'] 
        date = row['receivedAt']
        rgba = colormap(norm(speed))
        hex_color = colors.to_hex(rgba)
        popup_text = f"Name: {name}<br>Speed: {speed}<br>Date: {date}"
        folium.CircleMarker(
            location=[lat, lon],
            radius=5,
            color=hex_color,
            fill=True,
            fill_color=hex_color,
            fill_opacity=0.7,
            popup=folium.Popup(popup_text, max_width=200)
        ).add_to(m)

# Display the map
m
```

#### 3. Filter out stationary vessels

```python
# --- Parameters ---
REST_THRESHOLD_HOURS = 4       # total stopped time threshold
REST_THRESHOLD_SPEED = 1
MAX_GAP_MINUTES = 30           # cap per-message gap to avoid inflating stationary time

# Parse and sort
ais_df['receivedAt'] = pd.to_datetime(ais_df['receivedAt'], utc=True, errors='coerce')
ais_df['speedOverGround_kts'] = pd.to_numeric(ais_df['speedOverGround_kts'], errors='coerce')
ais_df = ais_df.sort_values(['mmsi', 'receivedAt'])

# Flag stopped rows
# ais_df['stopped'] = ais_df['speedOverGround_kts'].eq(0)
ais_df['stopped'] = ais_df['speedOverGround_kts'] <= REST_THRESHOLD_SPEED


# Time to next message for the same vessel
ais_df['next_time'] = ais_df.groupby('mmsi')['receivedAt'].shift(-1)
dt = (ais_df['next_time'] - ais_df['receivedAt']).dt.total_seconds()

# Not so sure about this part?? 
# Clean & cap gaps
# dt = dt.mask(dt < 0)  # drop negative intervals (if any ordering hiccups) --> was not needed
# dt_capped = dt.clip(upper=MAX_GAP_MINUTES * 60)

# Sum stationary time per vessel (in hours)
stopped_hours = (
    # dt_capped.where(ais_df['stopped'])
    dt.where(ais_df['stopped'])
    .groupby(ais_df['mmsi']).sum(min_count=1)
    .div(3600)
    .fillna(0)
)

# Vessels resting >= threshold
resting_vessels = stopped_hours[stopped_hours >= REST_THRESHOLD_HOURS].index

print("Total vessels:", ais_df['mmsi'].nunique())
# print(f"Vessels with total SOG=0 >= {REST_THRESHOLD_HOURS}h: {len(resting_vessels)}")
print(f"Vessels with SOG<={REST_THRESHOLD_SPEED}kts for >= {REST_THRESHOLD_HOURS}h: {len(resting_vessels)}")
print("Sample (top 10 by stopped hours):")
print(stopped_hours.sort_values(ascending=False).head(20))

# Filter them out
ais_active = ais_df[~ais_df['mmsi'].isin(resting_vessels)].copy()

print(f"Original messages: {len(ais_df)}")
print(f"Messages after removing resting vessels: {len(ais_active)}")
```
