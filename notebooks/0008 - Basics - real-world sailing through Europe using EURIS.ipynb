## Basic trip over a real-world graph
In this notebook, we set up a basic simulation where one vessel moves over another real-world graph. The vessel will sail from a location in the Maasvlakte, to a location near Duisburg, Germany.

```python
# package(s) used for creating and geo-locating the graph
import networkx as nx
import shapely

# package(s) related to the simulation (creating the vessel, running the simulation)
import datetime
import simpy
import opentnsim

# OpenTNSim mixins and utils
from opentnsim.core import Identifiable, Movable
from opentnsim.core.logutils import logbook2eventtable
from opentnsim.core.utils import create_object
from opentnsim.core.visualizations import generate_vessel_gantt_chart
import opentnsim.fis as fis
from opentnsim.graph.utils import get_closest_node_to_geometry
from opentnsim.graph.visualizations import plot_graph_folium, visualize_node_in_folium_plot, visualize_route_in_folium_plot

# package(s) needed for inspecting the output
import pandas as pd
import geopandas as gpd

print("This notebook is executed with OpenTNSim version {}".format(opentnsim.__version__))
```

#### 0. Create environment
We create the environment

```python
# start simpy environment
simulation_start = datetime.datetime(2024, 1, 1, 0, 0, 0)
env = simpy.Environment(initial_time=simulation_start.timestamp())
env.epoch = simulation_start
```

#### 1. Create graph
Next we need our graph. In this example we'll load the [EURIS graph](https://zenodo.org/records/17298014) from Zenodo.

```python
FG = fis.load_network(network="euris", version="0.1")
```

```python
maasvlakte = shapely.Point(4.0566946, 51.9471624)
duisburg = shapely.Point(6.722793, 51.451388)

# origin and destination node
origin_node = get_closest_node_to_geometry(FG, maasvlakte)
destination_node = get_closest_node_to_geometry(FG, duisburg)
```

```python
route = nx.shortest_path(FG, origin_node, destination_node, weight='length_m')
```

```python
m = plot_graph_folium(FG, longitude=5.38, latitude=51.75, zoom_start = 9)

visualize_route_in_folium_plot(m, FG, route, color='black', label='Route from Maasvlakte to Moerdijk')
visualize_node_in_folium_plot(m, FG, origin_node, color='lightgreen', label='Maasvlakte')
visualize_node_in_folium_plot(m, FG, destination_node, color='red', label='Moerdijk')

# Display the map
m
```

```python
env.graph = FG
```

#### 2. Create vessel
We create a vessel

```python
# make your preferred Vessel class out of available mix-ins.
Vessel = create_object(
    "Vessel", 
    (
        Identifiable, # allows to give the object a name and a random ID,
        Movable,      # allows the object to move, with a fixed speed, while logging this activity
    )
)
```

```python
def mission(env, vessel):
    """
    Method that defines the mission of the vessel. 
    
    In this case: 
        keep moving along the path until its end point is reached
    """
    while True:
        yield from vessel.move()
        
        if vessel.geometry == nx.get_node_attributes(env.graph, "geometry")[vessel.route[-1]]:
            break
```

```python
# create vessel from a dict 
path = route
data_vessel = {
    "env": env,                                       # needed for simpy simulation
    "name": "Vessel",                                 # required by Identifiable
    "geometry": env.graph.nodes[path[0]]['geometry'], # required by Locatable
    "route": path,                                    # required by Routeable
    "v": 1,                                           # required by Movable, 1 m/s to check if the distance is covered in the expected time
}  # 

# create an instance of the Vessel class using the input dict data_vessel
vessel = Vessel(**data_vessel)

# start the simulation
env.process(mission(env, vessel))
```

#### 3. Run simulation

```python
env.run()
```

# Show the logbook
Show all the logbook messages on a map. 

```python
logbook_gdf = gpd.GeoDataFrame(vessel.logbook, geometry="Geometry", crs="EPSG:4326")
logbook_gdf.explore()
```

```python
df_eventtable = logbook2eventtable([vessel])
duration = df_eventtable['duration (s)'].sum()
distance = df_eventtable['distance (m)'].sum()

print('{} sailed {:.1f} m in {:.1f} s, which is {}'.format(vessel.name, distance, duration, str(datetime.timedelta(seconds=duration))))
```
